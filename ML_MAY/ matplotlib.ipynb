## subplot

```python
x=[1,2,3,4,5,6]
y=[5,15,20,25,30,35]
y1=[i*5 for i in x]
y2=[i**2 for i in x]
y3=[i+5 for i in x]
print(x)
print(y)
print(y1)
print(y2)
print(y3)
```

```python
import matplotlib.pyplot as plt
```

```python

plt.subplot(2,2,1)
plt.plot(x,y,ls="-")

plt.subplot(2,2,2)
plt.plot(x,y1,ls="--")

plt.subplot(2,2,3)
plt.plot(x,y2,ls=":")

plt.subplot(2,2,4)
plt.plot(x,y3,ls="-.")
plt.show()
```

## pie chart

```python
student_values=[54,23,35,67]
student_performance=['Excellent','good','intermediate','poor']
plt.pie(student_values,labels=student_performance)
plt.show()
```

```python
plt.pie(student_values,labels=student_performance,explode=[0,0.1,0,0.2])
plt.show()
```

```python
plt.pie(student_values,labels=student_performance,explode=[0,0.1,0,0.2],shadow=True,startangle=90,autopct='%1.1f%%')
plt.show()
```

```python
plt.pie(student_values,labels=student_performance,explode=[0,0.1,0,0.2],shadow=True,startangle=90,autopct='%1.1f%%',counterclock=False)
plt.legend()
plt.show()
```
